```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def norma_complexa(zc:complex):
  return float(np.sqrt((zc*zc.conjugate()).real))
```

##Determinação de regição de estabilidade para Euler Exponencial

```python
def phi1(hc):
  if hc == 0:
    f = 1
    return f
  else:
    f = (np.exp(hc) - 1)/(hc)
    return f

def phi2(hc):
  if hc == 0:
    f = 0.5
    return f
  else:
    f = (np.exp(hc) - hc - 1)/(hc)**2
    return f

def fator_exp_1(z,ch):
  f = np.exp(ch) + z*phi1(ch)
  return f

def fator_exp_2_1(z,ch): #c_2 = 1
  f = np.exp(ch) + (phi1(ch) - phi2(ch))*z + phi2(ch)*z*(np.exp(ch) + phi1(ch)*z)
  return f

def fator_exp_2_2(z,ch): #c_2 = 1/2
  f = np.exp(ch) + (phi1(ch) - 2*phi2(ch))*z + 2*phi2(ch)*z*(np.exp(ch*0.5) + 0.5*phi1(0.5*ch)*z)
  return f
```

```python
def Metodo2(func,tol = 10**(-2)):
  N = 1000
  x = np.linspace(-10,10,N)
  y = np.linspace(-10,10,N)

  xcurl = np.array([])
  ycurl = np.array([])


  for i in range(N):
    for k in range(N):
      norma = norma_complexa(func(1j*x[i], 1j*y[k]))
      if norma > -1 and norma < 1 :
        xcurl = np.append(xcurl,x[i])
        ycurl = np.append(ycurl,y[k])

  return xcurl,ycurl
```

```python
met = fator_exp_1

x1,y1 = Metodo2(met)
```

```python
axxp = (0,0)
axxn = (10,-10)

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(x1,y1,color = 'b',s = 5)
ax.plot(axxn,axxp,color = 'k')
ax.plot(axxp,axxn,color = 'k')


plt.xlim(-10,10)
plt.ylim(-10,10)
plt.legend()
plt.title('Região de estabilidade - Euler Exponencial')
plt.ylabel('ch')
plt.xlabel('λh')
```
