```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def A1(alp,lamb):
  A = (1 - alp + alp*np.cos((2*np.pi)/lamb))**2 + (alp*np.sin((2*np.pi)/lamb))**2
  return A

def A2a(alp,lamb):
  re = (2-3*alp + alp**2)*0.5  + (2*alp-alp**2)*np.cos((2*np.pi)/lamb) + 0.5*(alp**2-alp)*np.cos(2*(2*np.pi)/lamb)
  im = (2*alp-alp**2)*np.sin((2*np.pi)/lamb) + 0.5*(alp**2-alp)*np.sin(2*(2*np.pi)/lamb)
  A = (re)**2 + (im)**2
  return A

def A2b(alp,lamb):
  A = (1 - alp**2 + (alp**2)*np.cos((2*np.pi)/lamb))**2 + (alp*np.sin((2*np.pi)/lamb))**2
  return A

def A3(alp,lamb):
  C1 = -alp*(2-alp)*(1-alp)/6
  C2 = (1-alp)*(2-alp)*(1+alp)/2
  C3 = alp*(2-alp)*(1+alp)/2
  C4 = alp*(alp-1)*(alp+1)/6

  re = C1*np.cos(2*np.pi/lamb) + C2 + C3*np.cos(2*np.pi/lamb) + C4*np.cos(2*2*np.pi/lamb)
  im = C1*np.sin(2*np.pi/lamb) - C3*np.sin(2*np.pi/lamb) - C4*np.sin(2*2*np.pi/lamb)

  A = re**2 + im**2
  return A
```

```python
alpha = np.linspace(0, 1, 1000)

a = np.zeros((3,len(alpha)))
for i in range(3):
  a[i,:] = A1(alpha,2**(i+1))


```

```python
plt.plot(alpha,a[0,:],color='r', label = 'λ = 2Δx')
plt.plot(alpha,a[1,:],color='g', label = 'λ = 4Δx')
plt.plot(alpha,a[2,:],color='b', label = 'λ = 8Δx')


plt.tick_params(axis='both', which='major', labelsize=12)

plt.title('|A1(α)|² vs α',fontsize = 16)
plt.xlabel('α',fontsize = 14)
plt.ylabel('|A(α)|²',fontsize = 14)
plt.legend()
plt.grid(True, which='both')
plt.show()
```

```python
def R1(p, lamb,alp= 0.5):
  alpha = alp
  R = ((lamb)/(2*np.pi*(p+alpha)))*((p*2*np.pi)/(lamb) + np.arctan2((alpha*np.sin(2*np.pi/lamb)),(1-alpha+alpha*np.cos(2*np.pi/lamb))))
  return R

def R2a(p,lamb,alp = 0.5):
  alpha = alp
  re = (2-3*alpha + alpha**2)*0.5  + (2*alpha-alpha**2)*np.cos((2*np.pi)/lamb) + 0.5*(alpha**2-alpha)*np.cos(2*(2*np.pi)/lamb)
  im = (2*alpha-alpha**2)*np.sin((2*np.pi)/lamb) + 0.5*(alpha**2-alpha)*np.sin(2*(2*np.pi)/lamb)
  cons = ((lamb)/(2*np.pi*(p+alpha)))*((p*2*np.pi)/(lamb))
  artan = ((lamb)/(2*np.pi*(p+alpha)))*np.arctan2((im),(re))
  R = cons + artan
  return R


def R2b(p,lamb,alp = 0.5):
  alpha = alp
  re = 1 - alpha**2 + (alpha**2)*np.cos((2*np.pi)/lamb)
  im = alpha*np.sin((2*np.pi)/lamb)
  cons = ((lamb)/(2*np.pi*(p+alpha)))*((p*2*np.pi)/(lamb))
  artan = ((lamb)/(2*np.pi*(p+alpha)))*np.arctan2((im),(re))
  R = cons + artan
  return R


def R3(p, lamb,alp = 0.5):
  alpha = alp

  C1 = -alpha*(2-alpha)*(1-alpha)/6
  C2 = (1-alpha)*(2-alpha)*(1+alpha)/2
  C3 = alpha*(2-alpha)*(1+alpha)/2
  C4 = alpha*(alpha-1)*(alpha+1)/6

  re = C1*np.cos(2*np.pi/lamb) + C2 + C3*np.cos(2*np.pi/lamb) + C4*np.cos(2*2*np.pi/lamb)
  im = C1*np.sin(2*np.pi/lamb) - C3*np.sin(2*np.pi/lamb) - C4*np.sin(2*2*np.pi/lamb)

  cons = ((lamb)/(2*np.pi*(p+alpha)))*((p*2*np.pi)/(lamb))
  artan = ((lamb)/(2*np.pi*(p+alpha)))*np.arctan2((-im),(re))

  R = cons + artan
  return R
```

```python
lamb = np.linspace(2.0000001, 10, 1000)
alp = 0.5
r = np.zeros((3,len(lamb)))
for i in range(3):
  r[i,:] = R1(i,lamb,alp = alp)


```

```python
plt.plot(lamb,r[0,:],color='r', label = 'p=0')
plt.plot(lamb,r[1,:],color='g', label = 'p=1')
plt.plot(lamb,r[2,:],color='b', label = 'p=2')
# subscrito a ₐ
# subscrito b

plt.tick_params(axis='both', which='major', labelsize=12)

plt.title('R1 vs λ - α = 0.5', fontsize = 16)
plt.xlabel('λ',fontsize = 14)
if alp ==0.5:
  plt.ylim(0,1.1)
plt.ylabel('R', fontsize = 14)
plt.legend()
plt.grid(True, which='both')
plt.show()
```

#Comparação do comportamento de R1 com R3 e A1 com A3

```python
alpha = np.linspace(0, 1, 1000)

a = np.zeros((2,len(alpha)))
r = np.zeros((2,len(alpha)))

a[0,:] = A1(alpha,8)
a[1,:] = A3(alpha,8)

lamb = np.linspace(2.0000001, 10, 1000)
r[0,:] = R1(0,lamb,alp=0.75)
r[1,:] = R3(0,lamb,alp=0.75)
```
